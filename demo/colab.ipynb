```python
import torch
import torch.nn as nn

# 1. Select the Device (Crucial step!)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# 2. Create a simple model and move it to GPU
model = nn.Linear(1, 1).to(device) 

# 3. Create dummy data and move it to GPU
# x is input, y is target (y = 2x + 1)
x_train = torch.tensor([[1.0], [2.0], [3.0]], device=device)
y_train = torch.tensor([[3.0], [5.0], [7.0]], device=device)

# 4. Define Loss and Optimizer
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 5. Training Loop
for epoch in range(100):
    # Forward pass
    outputs = model(x_train)
    loss = criterion(outputs, y_train)
    
    # Backward pass and optimization
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    if (epoch+1) % 20 == 0:
        print(f'Epoch [{epoch+1}/100], Loss: {loss.item():.4f}')

print("Training finished on GPU!")
```
